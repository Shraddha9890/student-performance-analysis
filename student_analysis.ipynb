```python

df.isnull().sum()
```

### Import Visualization Libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1) Gender vs Math Score

```python
sns.barplot(x='gender', y='math score', data=df)
plt.title("Gender vs Math Score")
plt.show()
```

### Male students scored slightly higher in math

# 2)Test Preparation Impact

```python
sns.barplot(x='test preparation course', y='math score', data=df)
plt.title("Test Preparation vs Math Score")
plt.show()
```

### Students who completed test preparation scored higher. Preparation has a positive impact on performance

# 3) Parental Education Impact

```python
sns.barplot(x='parental level of education', y='math score', data=df)
plt.xticks(rotation=45)
plt.show()
```

### Reading and writing scores are highly correlated. Math has moderate correlation with other subjects

# 4) Correlation Heatmap

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Between Subjects")
plt.show()
```

### Most students scored between 60–80 . Data is normally distributed

# 5) Score Distribution

```python
sns.histplot(df['math score'], kde=True)
plt.title("Math Score Distribution")
plt.show()
```

### Add Advanced feature

```python
df['average score'] = (df['math score'] + df['reading score'] + df['writing score']) / 3
```

```python
sns.boxplot(x='gender', y='average score', data=df)
plt.title("Gender vs Average Score")
plt.show()
```

## 🧠 Key Insights

- Students who completed test preparation scored significantly higher.
- Reading and writing scores show strong positive correlation.
- Parental education has a moderate impact on student performance.
- Female students performed better in reading and writing.
- Most students scored between 60–80 across subjects.
